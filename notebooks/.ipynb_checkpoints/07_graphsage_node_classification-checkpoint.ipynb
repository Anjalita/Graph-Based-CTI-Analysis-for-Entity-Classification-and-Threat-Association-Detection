# Notebook 7 — Final Robust GraphSAGE Evaluation

## Aim

This notebook performs the final Graph Neural Network evaluation for the Cyber Threat Intelligence project.

It addresses the limitations of the preliminary GraphSAGE run by using:

- ten random seeds and mean ± standard deviation;
- the exact canonical node features and split assignments created in Notebooks 5 and 6;
- a direction-preserving GraphSAGE architecture;
- LayerNorm instead of BatchNorm;
- validation-only hyperparameter selection;
- a complete full-graph experiment;
- a connected-node sensitivity experiment;
- repeated Logistic Regression and Random Forest baselines;
- an original-versus-canonical graph counterfactual;
- robust ROC-AUC calculation for subsets with missing classes;
- explicit overfitting, confidence, and error analysis.

## Prediction task

Each node is classified into its cybersecurity entity type, such as malware, threat actor, software, vulnerability, domain, or URL.

## Experimental interpretation

The **full canonical graph** is the main system-level result because it uses the same population and fixed split as Notebook 6.

The **connected-node experiment** is the main graph-effect sensitivity analysis because GraphSAGE can aggregate neighbourhood information only for connected nodes.

The **original-versus-canonical experiment** is a pipeline-level counterfactual. Canonicalisation changes the number and identity of nodes, so it is not treated as a perfectly matched-node causal experiment.

## 6. Preserve edge direction

The original CTI graph is directed. This notebook does not symmetrise its edges.

Each GraphSAGE layer separately aggregates:

- incoming-neighbour information using the original edge direction;
- outgoing-neighbour information using reversed edges.

The two representations are combined before classification.

```python
#Keeps A → B different from B → A so CTI relationship direction is preserved
def graph_to_edge_index(
    graph: nx.Graph,
    node_to_index: dict[str, int],
    condition_name: str
) -> tuple[torch.Tensor, dict]:

    total_graph_edges= graph.number_of_edges()
    mapped_edges= []
    dropped_edges= 0

    for source, target in graph.edges():
        source_key= str(source)
        target_key= str(target)

        if (
            source_key in node_to_index
            and target_key in node_to_index ):
            mapped_edges.append([
                node_to_index[source_key],
                node_to_index[target_key]])
        else:
            dropped_edges += 1

    if not mapped_edges:
        raise ValueError(f"No edges could be mapped for {condition_name}.")

    raw_edge_index = torch.tensor(
        mapped_edges,
        dtype=torch.long
    ).t().contiguous()

    raw_mapped_edges = raw_edge_index.shape[1]
    raw_self_loops = int(
        (raw_edge_index[0] == raw_edge_index[1]).sum()
    )

    # SAGEConv already contains a root transformation, so explicit
    # self-loops are removed to avoid double-counting the node itself.
    edge_index, _ = remove_self_loops(
        raw_edge_index
    )

    # Remove duplicate directed edges while preserving direction.
    edge_index = coalesce(
        edge_index,
        num_nodes=len(node_to_index) )

    diagnostics = {
        "condition": condition_name,
        "total_graph_edges": int(total_graph_edges),
        "mapped_edges_before_cleaning": int(raw_mapped_edges),
        "dropped_edges_unmapped_endpoint": int(dropped_edges),
        "self_loops_removed": int(raw_self_loops),
        "unique_directed_edges_retained": int(
            edge_index.shape[1] ),
        "edge_retention_percentage": float(
            100 * raw_mapped_edges / max(total_graph_edges, 1) ) }

    return edge_index, diagnostics


canonical_node_to_index = dict(
    zip(
        canonical_embeddings_df[
            "node_id"
        ].astype(str),
        canonical_embeddings_df[
            "node_index"
        ].astype(int)))

(canonical_edge_index,canonical_edge_diagnostics)= graph_to_edge_index(canonical_graph,canonical_node_to_index, "canonical_graph")

canonical_reverse_edge_index = (canonical_edge_index.flip(0))

canonical_edge_diagnostics_df = pd.DataFrame([canonical_edge_diagnostics])

display(canonical_edge_diagnostics_df)
```

## 8. Connected-node eligibility rule

Rare entity classes may have no connected training examples or only one connected validation/test example. Such classes cannot support a valid connected-only supervised experiment.

A class is eligible for the connected sensitivity analysis only when it has:

- at least 3 connected training nodes;
- at least 1 connected validation node;
- at least 1 connected test node.

No rows are silently discarded: eligible and excluded classes are saved and reported.

```python
class_eligibility_records = []
#Selects classes with enough connected nodes for a reliable connected-only experiment
for encoded_label, entity_type in zip(
    label_mapping_df["encoded_label"],
    label_mapping_df["entity_type"]):
    label_mask = canonical_y.eq(
        int(encoded_label))

    train_count = int(
        (
            label_mask
            & train_mask
            & connected_mask
        ).sum()
    )

    validation_count = int(
        (
            label_mask
            & validation_mask
            & connected_mask
        ).sum()
    )

    test_count = int(
        (
            label_mask
            & test_mask
            & connected_mask
        ).sum()
    )

    eligible = (
        train_count >= 3
        and validation_count >= 1
        and test_count >= 1)

    class_eligibility_records.append({
        "encoded_label": int(encoded_label),
        "entity_type": entity_type,
        "connected_train": train_count,
        "connected_validation": validation_count,
        "connected_test": test_count,
        "eligible": eligible})

connected_class_eligibility_df = pd.DataFrame(class_eligibility_records)

eligible_connected_labels = set(
    connected_class_eligibility_df.loc[
        connected_class_eligibility_df[
            "eligible"
        ],
        "encoded_label"
    ].astype(int)
)

eligible_label_mask = torch.tensor(
    [
        int(label) in eligible_connected_labels
        for label in canonical_y.tolist()
    ],
    dtype=torch.bool)

connected_train_mask = (
    train_mask
    & connected_mask
    & eligible_label_mask)

connected_validation_mask = (
    validation_mask
    & connected_mask
    & eligible_label_mask)

connected_test_mask = (
    test_mask
    & connected_mask
    & eligible_label_mask)

display(connected_class_eligibility_df)

print(
    "Eligible connected classes:",
    len(eligible_connected_labels))

print(
    "Connected eligible train/validation/test:",
    int(connected_train_mask.sum()),
    int(connected_validation_mask.sum()),
    int(connected_test_mask.sum()))
```

## 10. Robust metrics

ROC-AUC is calculated only for one-vs-rest class comparisons that contain both positive and negative cases. This prevents invalid subgroup AUC values when a class is absent.

```python
def robust_multiclass_auc(
    y_true: np.ndarray,
    probabilities: np.ndarray,
    num_classes: int
) -> tuple[float, float, int]:

    auc_values = []
    class_supports = []

    for class_id in range(num_classes):
        binary_target = (
            y_true == class_id).astype(int)

        if (
            binary_target.min()
            == binary_target.max()):
            continue

        auc_values.append(
            roc_auc_score(
                binary_target,
                probabilities[:, class_id]))

        class_supports.append(
            binary_target.sum())

    if not auc_values:
        return np.nan, np.nan, 0

    macro_auc = float(
        np.mean(auc_values))

    weighted_auc = float(
        np.average(
            auc_values,
            weights=class_supports))

    return (macro_auc,weighted_auc,len(auc_values))


def calculate_metrics(
    y_true: np.ndarray,
    y_pred: np.ndarray,
    probabilities: np.ndarray | None = None) -> dict:

    present_labels = np.unique(y_true)

    metrics = {
        "accuracy": accuracy_score(
            y_true,
            y_pred),
        "balanced_accuracy": (
            balanced_accuracy_score(
                y_true,
                y_pred)),
        "macro_precision": precision_score(
            y_true,
            y_pred,
            labels=present_labels,
            average="macro",
            zero_division=0),
        "macro_recall": recall_score(
            y_true,
            y_pred,
            labels=present_labels,
            average="macro",
            zero_division=0),
        "macro_f1": f1_score(
            y_true,
            y_pred,
            labels=present_labels,
            average="macro",
            zero_division=0),
        "weighted_precision": precision_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0),
        "weighted_recall": recall_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        ),
        "weighted_f1": f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        ),
        "mcc": matthews_corrcoef(
            y_true,
            y_pred
        ),
        "macro_metric_classes": int(
            len(present_labels)
        )
    }

    if probabilities is not None:
        (
            macro_auc,
            weighted_auc,
            evaluated_classes
        ) = robust_multiclass_auc(
            y_true,
            probabilities,
            probabilities.shape[1]
        )

        metrics[
            "macro_roc_auc_present_classes"
        ] = macro_auc

        metrics[
            "weighted_roc_auc_present_classes"
        ] = weighted_auc

        metrics[
            "auc_evaluated_classes"
        ] = evaluated_classes

    return metrics
```

## 11. Direction-preserving GraphSAGE with LayerNorm

```python
#Defines the GraphSAGE neural network that separately learns incoming and outgoing relationship
class DirectedGraphSAGELayer(nn.Module):
    def __init__(
        self,
        input_dim: int,
        output_dim: int
    ):
        super().__init__()

        self.incoming = SAGEConv(
            input_dim,
            output_dim
        )

        self.outgoing = SAGEConv(
            input_dim,
            output_dim
        )

        self.combine = nn.Linear(
            output_dim * 2,
            output_dim
        )

    def forward(
        self,
        x: torch.Tensor,
        edge_index: torch.Tensor,
        reverse_edge_index: torch.Tensor
    ) -> torch.Tensor:

        incoming_x = self.incoming(x,edge_index)

        outgoing_x = self.outgoing(x,reverse_edge_index )

        return self.combine(torch.cat([incoming_x, outgoing_x],dim=1))
class DirectedGraphSAGEClassifier(nn.Module):
    def __init__(
        self,
        input_dim: int,
        hidden_dim: int,
        output_dim: int,
        dropout: float):
        super().__init__()

        self.layer1 = DirectedGraphSAGELayer(input_dim,hidden_dim)

        self.normalisation = nn.LayerNorm(hidden_dim)

        self.layer2 = DirectedGraphSAGELayer(hidden_dim,output_dim )
        self.dropout = dropout
    def forward(
        self,
        x: torch.Tensor,
        edge_index: torch.Tensor,
        reverse_edge_index: torch.Tensor
    ) -> torch.Tensor:

        x= self.layer1(x,edge_index,reverse_edge_index)

        x= self.normalisation(x)
        x= F.relu(x)

        x= F.dropout(x,p=self.dropout,training=self.training )

        return self.layer2(x,edge_index,reverse_edge_index)
```

## 12. Training and prediction utilities

### Class weighting

For each class \(c\) present in the active training population, the loss weight is:

\[
w_c = \sqrt{\frac{N}{K n_c}}
\]

where \(N\) is the number of active training nodes, \(K\) is the number of represented classes, and \(n_c\) is the class frequency. The square root tempers extreme inverse-frequency weights, after which the weights are mean-normalised.


```python
def make_class_weights(
    labels: torch.Tensor,
    active_train_mask: torch.Tensor,
    num_classes: int
) -> torch.Tensor:
#Creates reusable functions for training GraphSAGE and making predictions
    counts= torch.bincount(
        labels[active_train_mask].cpu(),minlength=num_classes).float()

    weights= torch.zeros_like(counts)

    present= counts > 0

    weights[present] = torch.sqrt(counts[present].sum()/ (present.sum()* counts[present]))

    if weights[present].numel() > 0:
        weights[present] = (weights[present]/ weights[present].mean())

    return weights.to(DEVICE)


@torch.no_grad()
def evaluate_graphsage(
    model: nn.Module,
    data: Data,
    reverse_edge_index: torch.Tensor,
    evaluation_mask: torch.Tensor
):
    model.eval()

    logits= model(data.x,data.edge_index,reverse_edge_index)

    selected_logits = logits[evaluation_mask]

    probabilities = torch.softmax(selected_logits,dim=1 ).cpu().numpy()

    predictions = selected_logits.argmax(dim=1 ).cpu().numpy()

    true_labels = data.y[evaluation_mask].cpu().numpy()

    metrics = calculate_metrics(
        true_labels,
        predictions,
        probabilities)

    return (
        metrics,
        true_labels,
        predictions,
        probabilities
    )


def train_graphsage_once(
    data: Data,
    reverse_edge_index: torch.Tensor,
    active_train_mask: torch.Tensor,
    active_validation_mask: torch.Tensor,
    seed: int,
    config: dict,
    return_model: bool = False
):
    set_seed(seed)

    model = DirectedGraphSAGEClassifier(
        input_dim=data.x.shape[1],
        hidden_dim=config[
            "hidden_dim"
        ],
        output_dim=NUM_CLASSES,
        dropout=config["dropout"]
    ).to(DEVICE)

    optimiser = torch.optim.AdamW(
        model.parameters(),
        lr=config["learning_rate"],
        weight_decay=config[
            "weight_decay"
        ]
    )

    class_weights = make_class_weights(
        data.y,
        active_train_mask,
        NUM_CLASSES
    )

    loss_function = nn.CrossEntropyLoss(
        weight=class_weights
    )

    best_model_state = None
    best_validation_macro_f1 = -np.inf
    best_validation_loss = np.inf
    best_epoch = 0

    epochs_without_loss_improvement = 0
    history_records = []

    start_time = time.time()

    for epoch in range(
        1,
        config["maximum_epochs"] + 1
    ):
        model.train()
        optimiser.zero_grad()

        logits = model(
            data.x,
            data.edge_index,
            reverse_edge_index
        )

        training_loss = loss_function(
            logits[active_train_mask],
            data.y[active_train_mask]
        )

        training_loss.backward()
        optimiser.step()

        model.eval()

        with torch.no_grad():
            validation_logits = model(
                data.x,
                data.edge_index,
                reverse_edge_index
            )

            validation_loss = loss_function(
                validation_logits[
                    active_validation_mask
                ],
                data.y[
                    active_validation_mask
                ]
            ).item()

        (
            training_metrics,
            _,
            _,
            _
        ) = evaluate_graphsage(
            model,
            data,
            reverse_edge_index,
            active_train_mask
        )

        (
            validation_metrics,
            _,
            _,
            _
        ) = evaluate_graphsage(
            model,
            data,
            reverse_edge_index,
            active_validation_mask
        )

        history_records.append({
            "epoch": epoch,
            "training_loss": float(
                training_loss.item()
            ),
            "validation_loss": float(
                validation_loss
            ),
            "training_macro_f1": (
                training_metrics[
                    "macro_f1"
                ]
            ),
            "validation_macro_f1": (
                validation_metrics[
                    "macro_f1"
                ]
            )
        })

        if (epoch>= config["minimum_epochs"]
        ):
            if (
                validation_metrics[
                    "macro_f1"
                ]
                > best_validation_macro_f1
                + 1e-6
            ):
                best_validation_macro_f1 = (
                    validation_metrics[
                        "macro_f1"
                    ]
                )

                best_model_state = copy.deepcopy(
                    model.state_dict()
                )

                best_epoch = epoch

            if (
                validation_loss
                < best_validation_loss
                - 1e-5
            ):
                best_validation_loss = (
                    validation_loss
                )

                epochs_without_loss_improvement = 0

            else:
                epochs_without_loss_improvement += 1

            if (
                epochs_without_loss_improvement
                >= config[
                    "early_stopping_patience"
                ]
            ):
                break

    if best_model_state is None:
        best_model_state = copy.deepcopy(
            model.state_dict()
        )

        best_epoch = len(
            history_records
        )

    model.load_state_dict(
        best_model_state
    )

    result = {
        "history": pd.DataFrame(
            history_records
        ),
        "best_epoch": best_epoch,
        "best_validation_macro_f1": (
            best_validation_macro_f1
        ),
        "training_time_seconds": (
            time.time() - start_time
        )
    }

    if return_model:
        result["model"] = model

    return result
```

## 13. Validation-only hyperparameter selection

The grid is intentionally small. Three tuning seeds are used, and the held-out test set is not examined.

```python
#Finds good GraphSAGE settings using validation data without touching test data
parameter_grid = {
    "hidden_dim": [16, 32, 64],
    "dropout": [0.30, 0.50],
    "learning_rate": [
        0.001,
        0.005
    ],
    "weight_decay": [5e-4],
    "maximum_epochs": [250],
    "minimum_epochs": [25],
    "early_stopping_patience": [25]}

tuning_records = []

for configuration in ParameterGrid(parameter_grid):
    validation_scores = []

    for seed in TUNING_SEEDS:
        run = train_graphsage_once(
            data=canonical_data,
            reverse_edge_index=(canonical_reverse_edge_index),
            active_train_mask=(train_mask.to(DEVICE)),
            active_validation_mask=(
                validation_mask.to(DEVICE)),
            seed=seed,
            config=configuration)

        validation_scores.append(
            run["best_validation_macro_f1"])

    tuning_records.append({
        **configuration,
        "validation_macro_f1_mean": float(
            np.mean(validation_scores)),
        "validation_macro_f1_std": float(
            np.std(
                validation_scores,
                ddof=1))})

tuning_results_df = pd.DataFrame(
    tuning_records
).sort_values(
    [
        "validation_macro_f1_mean",
        "validation_macro_f1_std"
    ],ascending=[False, True]).reset_index(drop=True)

best_configuration = {
    key: tuning_results_df.loc[
        0,
        key
    ]
    for key in parameter_grid
}

for key in [
    "hidden_dim",
    "maximum_epochs",
    "minimum_epochs",
    "early_stopping_patience"
]:
    best_configuration[key] = int(
        best_configuration[key]
    )

print("Selected GraphSAGE configuration:")
print(best_configuration)
display(tuning_results_df)
```

## 14. Multi-seed GraphSAGE experiment runner

```python
def run_graphsage_seeds(
    experiment_name: str,
    data: Data,
    reverse_edge_index: torch.Tensor,
    active_train_mask: torch.Tensor,
    active_validation_mask: torch.Tensor,
    active_test_mask: torch.Tensor,
    configuration: dict
):
    records = []
    histories = {}
    model_states = {}

    for seed in SEEDS:
        run = train_graphsage_once(
            data=data,
            reverse_edge_index=(
                reverse_edge_index
            ),
            active_train_mask=(
                active_train_mask
            ),
            active_validation_mask=(
                active_validation_mask
            ),
            seed=seed,
            config=configuration,
            return_model=True
        )

        (
            metrics,
            _,
            _,
            _
        ) = evaluate_graphsage(
            run["model"],
            data,
            reverse_edge_index,
            active_test_mask
        )

        records.append({
            "experiment": (
                experiment_name
            ),
            "seed": seed,
            "best_epoch": run[
                "best_epoch"
            ],
            "best_validation_macro_f1": (
                run[
                    "best_validation_macro_f1"
                ]
            ),
            "training_time_seconds": (
                run[
                    "training_time_seconds"
                ]
            ),
            **metrics
        })

        histories[seed] = run[
            "history"
        ]

        model_states[seed] = (
            copy.deepcopy(
                run["model"].state_dict()
            )
        )

    return (
        pd.DataFrame(records),
        histories,
        model_states
    )
```

## 17. Aggregate GraphSAGE mean ± standard deviation

```python
#Combines all runs to report average performance and variation
REPORTED_METRICS = [
    "accuracy",
    "balanced_accuracy",
    "macro_precision",
    "macro_recall",
    "macro_f1",
    "weighted_f1",
    "mcc",
    "macro_roc_auc_present_classes"
]


def summarise_seed_results(
    results_df: pd.DataFrame,
    experiment_name: str
) -> pd.DataFrame:

    summary = {
        "experiment": experiment_name,
        "runs": len(results_df),
        "best_epoch_mean": (
            results_df[
                "best_epoch"
            ].mean()
        ),
        "best_epoch_std": (
            results_df[
                "best_epoch"
            ].std(ddof=1)
        )
    }

    for metric in REPORTED_METRICS:
        summary[
            f"{metric}_mean"
        ] = results_df[
            metric
        ].mean()

        summary[
            f"{metric}_std"
        ] = results_df[
            metric
        ].std(ddof=1)

    return pd.DataFrame([summary])


full_graphsage_summary_df = (
    summarise_seed_results(
        full_graphsage_runs_df,
        "Canonical full graph"
    )
)

connected_graphsage_summary_df = (
    summarise_seed_results(
        connected_graphsage_runs_df,
        "Canonical connected eligible"
    )
)

graphsage_summary_df = pd.concat(
    [
        full_graphsage_summary_df,
        connected_graphsage_summary_df
    ],
    ignore_index=True
)

display(graphsage_summary_df)
```

## 18. Repeated traditional baselines

Logistic Regression and Random Forest are rerun across the same ten seeds.

For the full-graph comparison, the fixed Notebook 6 split is reused.

For the connected comparison, the same connected eligible-class masks are used.

The traditional models are not topology-free baselines because Node2Vec already encodes graph structure. The comparison therefore tests whether supervised neighbourhood aggregation adds value beyond fixed random-walk embeddings. No raw non-graph node attributes are available in the current dataset for a genuinely topology-free feature baseline.


```python
#Reruns Logistic Regression and Random Forest under the same conditions for comparison
def aligned_probabilities(
    model,
    feature_matrix: np.ndarray,
    num_classes: int
) -> np.ndarray:

    raw_probabilities = model.predict_proba(
        feature_matrix
    )

    aligned = np.zeros(
        (
            len(feature_matrix),
            num_classes
        ),
        dtype=float
    )

    for position, class_id in enumerate(
        model.classes_
    ):
        aligned[
            :,
            int(class_id)
        ] = raw_probabilities[
            :,
            position
        ]

    return aligned


def run_baseline_seeds(
    experiment_name: str,
    feature_matrix: np.ndarray,
    labels: np.ndarray,
    active_train_mask: np.ndarray,
    active_test_mask: np.ndarray
) -> pd.DataFrame:

    records = []

    x_train = feature_matrix[active_train_mask]

    y_train = labels[active_train_mask]

    x_test = feature_matrix[active_test_mask]

    y_test = labels[active_test_mask]

    for seed in SEEDS:
        models = {
            "Majority Class": DummyClassifier(
                strategy="most_frequent"),
            "Logistic Regression": LogisticRegression(
                max_iter=3000,
                class_weight="balanced",
                random_state=seed,
                solver="lbfgs"),
            "Random Forest": RandomForestClassifier(
                n_estimators=500,
                class_weight="balanced_subsample",
                random_state=seed,
                n_jobs=-1 )}

        for model_name, model in models.items():
            model.fit(
                x_train,
                y_train
            )

            predictions = model.predict(
                x_test
            )

            if hasattr(
                model,
                "predict_proba"
            ):
                probabilities = aligned_probabilities(
                    model,
                    x_test,
                    NUM_CLASSES
                )
            else:
                probabilities = None

            metrics = calculate_metrics(
                y_test,
                predictions,
                probabilities
            )

            records.append({
                "experiment": (
                    experiment_name
                ),
                "model": model_name,
                "seed": seed,
                **metrics
            })

    return pd.DataFrame(records)


feature_matrix_np = canonical_x.numpy()
labels_np = canonical_y.numpy()

full_baseline_runs_df = (
    run_baseline_seeds(
        experiment_name=(
            "canonical_full_graph"
        ),
        feature_matrix=feature_matrix_np,
        labels=labels_np,
        active_train_mask=(
            train_mask.numpy()
        ),
        active_test_mask=(
            test_mask.numpy()
        )
    )
)

connected_baseline_runs_df = (
    run_baseline_seeds(
        experiment_name=(
            "canonical_connected_eligible"
        ),
        feature_matrix=feature_matrix_np,
        labels=labels_np,
        active_train_mask=(
            connected_train_mask.numpy()
        ),
        active_test_mask=(
            connected_test_mask.numpy()
        )
    )
)

display(full_baseline_runs_df.head())
display(connected_baseline_runs_df.head())
```

## 19. Fair model comparison

```python

def summarise_model_runs(
    baseline_runs_df: pd.DataFrame,
    graphsage_runs_df: pd.DataFrame,
    experiment_name: str
) -> pd.DataFrame:

    graphsage_rows = (graphsage_runs_df.copy())

    graphsage_rows["model"]= "GraphSAGE"

    combined = pd.concat(
        [
            baseline_runs_df,
            graphsage_rows
        ],
        ignore_index=True,
        sort=False
    )

    aggregation = {}

    for metric in REPORTED_METRICS:
        aggregation[
            f"{metric}_mean"
        ] = (
            metric,
            "mean"
        )

        aggregation[f"{metric}_std"] = (metric,"std")

    summary = (
        combined
        .groupby("model")
        .agg(**aggregation)
        .reset_index())

    summary.insert(
        0,
        "experiment",
        experiment_name
    )

    return summary


full_model_comparison_df = (
    summarise_model_runs(
        full_baseline_runs_df,
        full_graphsage_runs_df,
        "Canonical full graph"
    )
)

connected_model_comparison_df = (
    summarise_model_runs(
        connected_baseline_runs_df,
        connected_graphsage_runs_df,
        "Canonical connected eligible"
    )
)

display(full_model_comparison_df)
display(connected_model_comparison_df)
```

## 20. Extract labelled graph nodes robustly

```python

def infer_entity_type(
    node_attributes: dict
) -> str | None:

    for key in [
        "entity_type",
        "type",
        "label",
        "node_type"
    ]:
        value = node_attributes.get(
            key
        )

        if (
            value is not None
            and str(value).strip()
        ):
            return str(value)

    return None


def prepare_graph_condition(
    graph: nx.Graph,
    condition_name: str
):
    records = []

    for (
        node_id,
        node_attributes
    ) in graph.nodes(data=True):

        entity_type = infer_entity_type(node_attributes)

        if entity_type in (entity_type_to_label):
            records.append({
                "node_id": str(node_id),
                "entity_type": entity_type,
                "encoded_label": int(
                    entity_type_to_label[
                        entity_type])})

    node_df= (
        pd.DataFrame(records)
        .drop_duplicates("node_id")
        .reset_index(drop=True)
    )

    node_df["node_index"]= np.arange(len(node_df))

    mapping = dict(
        zip(
            node_df["node_id"],
            node_df["node_index"]))
    edge_index, diagnostics = graph_to_edge_index(graph,mapping,condition_name)

    connected_nodes= set(torch.unique(edge_index).tolist())

    node_df[
        "connected"
    ] = node_df[
        "node_index"
    ].isin(connected_nodes)

    diagnostics.update({
        "total_graph_nodes": int(graph.number_of_nodes()),
        "labelled_nodes_retained": int(len(node_df)),
        "unlabelled_or_unrecognised_nodes_excluded": int(
            graph.number_of_nodes() - len(node_df)
        ),
        "connected_labelled_nodes": int(
            node_df["connected"].sum())})

    return (
        node_df,
        edge_index,
        diagnostics
    )
```

## 23. Deterministic class-wise split

```python
#Creates consistent train/validation/test splits for the comparison
def create_classwise_masks(
    node_df: pd.DataFrame,
    eligible_types: set[str],
    seed: int
):
    random_generator = np.random.default_rng(seed)

    train_indices = []
    validation_indices = []
    test_indices = []

    eligible_rows = node_df[
        node_df["connected"]
        & node_df["entity_type"].isin(
            eligible_types
        )
    ]

    for entity_type in sorted(
        eligible_types
    ):
        indices = eligible_rows.loc[
            eligible_rows[
                "entity_type"
            ].eq(entity_type),
            "node_index"
        ].to_numpy(copy=True)

        random_generator.shuffle(
            indices
        )

        count = len(indices)

        test_count = max(
            1,
            int(round(count * 0.15))
        )

        validation_count = max(
            1,
            int(round(count * 0.15))
        )

        if (
            count
            - test_count
            - validation_count
            < 3
        ):
            raise ValueError(
                f"Class {entity_type} does not "
                "have enough rows after splitting."
            )

        test_indices.extend(
            indices[:test_count]
        )

        validation_indices.extend(
            indices[
                test_count:
                test_count
                + validation_count
            ]
        )

        train_indices.extend(
            indices[
                test_count
                + validation_count:
            ]
        )

    train_mask_array = np.zeros(
        len(node_df),
        dtype=bool
    )

    validation_mask_array = np.zeros(
        len(node_df),
        dtype=bool
    )

    test_mask_array = np.zeros(
        len(node_df),
        dtype=bool
    )

    train_mask_array[
        train_indices
    ] = True

    validation_mask_array[
        validation_indices
    ] = True

    test_mask_array[
        test_indices
    ] = True

    return (
        torch.tensor(
            train_mask_array,
            dtype=torch.bool
        ),
        torch.tensor(
            validation_mask_array,
            dtype=torch.bool
        ),
        torch.tensor(
            test_mask_array,
            dtype=torch.bool
        )
    )
```

## 24. Fresh Node2Vec embeddings for both graph conditions

The parameters match the established Notebook 5 design: 128 dimensions and 50 epochs. Results are cached, so each graph is trained only once unless the cache is deleted.

```python
#Generates equivalent Node2Vec features for the original and canonical graphs
NODE2VEC_CONFIG = {
    "embedding_dim": 128,       # keep consistent with Notebook 5
    "walk_length": 10,          # reduced from 20
    "context_size": 5,          # reduced from 10
    "walks_per_node": 3,        # reduced from 10
    "p": 1.0,
    "q": 1.0,
    "num_negative_samples": 2,  # reduced from 5
    "batch_size": 512,          # increased from 128
    "learning_rate": 0.01,
    "epochs": 15,               # reduced from 50
    "seed": 42}


def train_or_load_node2vec(
    condition_name: str,
    edge_index: torch.Tensor,
    num_nodes: int
) -> np.ndarray:

    cache_file = (CACHE_DIR/ f"{condition_name}_node2vec.npy" )

    if cache_file.exists():
        print("Loading cached embeddings:",cache_file.name)

        return np.load(cache_file)

    set_seed(NODE2VEC_CONFIG["seed"])

    node2vec_edge_index = to_undirected(edge_index,num_nodes=num_nodes)

    model= Node2Vec(edge_index=(node2vec_edge_index),
        embedding_dim=(
            NODE2VEC_CONFIG[
                "embedding_dim"]),
        walk_length=(
            NODE2VEC_CONFIG["walk_length"]),
        context_size=(NODE2VEC_CONFIG["context_size"]),
        walks_per_node=(
            NODE2VEC_CONFIG[
                "walks_per_node"]),
        p=NODE2VEC_CONFIG["p"],
        q=NODE2VEC_CONFIG["q"],
        num_negative_samples=(NODE2VEC_CONFIG["num_negative_samples" ]),
        num_nodes=num_nodes,
        sparse=True
    ).to(DEVICE)

    loader = model.loader(
        batch_size=(
            NODE2VEC_CONFIG[
                "batch_size"]),
        shuffle=True,
        num_workers=0 )

    optimiser = torch.optim.SparseAdam(
        model.parameters(),
        lr=NODE2VEC_CONFIG[
            "learning_rate"])

    for epoch in range(
        1,
        NODE2VEC_CONFIG[
            "epochs"
        ] + 1
    ):
        model.train()
        epoch_loss = 0.0
        batches = 0

        for (
            positive_walk,
            negative_walk
        ) in loader:
            optimiser.zero_grad()

            loss = model.loss(
                positive_walk.to(DEVICE),
                negative_walk.to(DEVICE))

            loss.backward()
            optimiser.step()

            epoch_loss += loss.item()
            batches += 1

        if (
            epoch == 1
            or epoch % 10 == 0
        ):
            print(
                condition_name,
                "| epoch",
                epoch,
                "| loss",
                round(
                    epoch_loss
                    / max(batches, 1),
                    4 ))

    embeddings = (
        model.embedding.weight
        .detach()
        .cpu()
        .numpy())

    np.save(cache_file,embeddings)

    return embeddings
```

## 28. Representative diagnostics

A representative seed is selected as the run closest to the ten-seed full-graph mean. It is used only for plots and examples. Statistical claims are based on the aggregate results.

```python
#Chooses one typical run only for plots and detailed inspection
full_macro_f1_mean = (
    full_graphsage_runs_df[
        "macro_f1"
    ].mean()
)

representative_seed = int(
    full_graphsage_runs_df.loc[
        (
            full_graphsage_runs_df[
                "macro_f1"
            ]
            - full_macro_f1_mean
        ).abs().idxmin(),
        "seed"
    ]
)

representative_model = (
    DirectedGraphSAGEClassifier(
        input_dim=canonical_data.x.shape[1],
        hidden_dim=best_configuration[
            "hidden_dim"
        ],
        output_dim=NUM_CLASSES,
        dropout=best_configuration[
            "dropout"
        ]
    ).to(DEVICE)
)

representative_model.load_state_dict(
    full_graphsage_states[
        representative_seed
    ]
)

(
    representative_metrics,
    representative_true,
    representative_predictions,
    representative_probabilities
) = evaluate_graphsage(
    representative_model,
    canonical_data,
    canonical_reverse_edge_index,
    test_mask.to(DEVICE)
)

print(
    "Representative seed:",
    representative_seed
)

print(representative_metrics)
```

## 29. Training curve and overfitting evidence

```python
#Checks whether GraphSAGE learns properly or starts overfittin
representative_history_df = (
    full_graphsage_histories[
        representative_seed])

plt.figure(figsize=(10, 6))

plt.plot(representative_history_df[ "epoch"],
    representative_history_df[
        "training_macro_f1"],
    label="Training macro F1")

plt.plot(representative_history_df["epoch"],
    representative_history_df[
        "validation_macro_f1"],
    label="Validation macro F1")

plt.xlabel("Epoch")
plt.ylabel("Macro F1")
plt.title(
    "GraphSAGE Training and Validation "
    f"Performance — Seed {representative_seed}")
plt.legend()
plt.tight_layout()

training_curve_path = (
    FIGURE_DIR
    / "graphsage_training_validation_curve.png")

plt.savefig(
    training_curve_path,
    dpi=300,
    bbox_inches="tight")

plt.show()
print("Starting hyperparameter tuning...")
```

## 30. Confusion matrix, confidence, and error analysis

```python
#Examines which entity classes are confused and how confident predictions are
representative_confusion = confusion_matrix(
    representative_true,
    representative_predictions,
    labels=np.arange(NUM_CLASSES))

fig, ax = plt.subplots(figsize=(14, 12))

ConfusionMatrixDisplay(
    confusion_matrix=(
        representative_confusion
    ),
    display_labels=class_names
).plot(
    ax=ax,
    xticks_rotation=90,
    values_format="d",
    colorbar=False)

plt.title("GraphSAGE Full-Test Confusion Matrix "f"— Representative Seed {representative_seed}")

plt.tight_layout()

confusion_matrix_path= (FIGURE_DIR/ "graphsage_full_test_confusion_matrix.png")

plt.savefig(
    confusion_matrix_path,
    dpi=300,
    bbox_inches="tight")

plt.show()

confidence_values= (representative_probabilities.max(axis=1))

confidence_summary_df = pd.DataFrame([{
    "mean_confidence": float(
        confidence_values.mean()),
    "median_confidence": float(
        np.median(confidence_values)),
    "minimum_confidence": float(
        confidence_values.min()),
    "maximum_confidence": float(
        confidence_values.max()),
    "uniform_20_class_reference": float( 1 / NUM_CLASSES)}])

display(confidence_summary_df)

test_rows = np.where(
    test_mask.numpy())[0]

prediction_columns = [
    "node_index",
    "node_id",
    "entity_type",
    "canonical_text",
    "mention_count",
    "degree",
    "encoded_label",
    "split"]

prediction_df = (
    canonical_split_df.iloc[
        test_rows
    ]
    .reindex(
        columns=prediction_columns
    )
    .copy()
)

prediction_df[
    "predicted_encoded_label"
] = representative_predictions

prediction_df[
    "predicted_entity_type"
] = [
    class_names[label]
    for label in representative_predictions]

prediction_df[
    "correct_prediction"
] = (
    prediction_df[
        "encoded_label"
    ].to_numpy()
    == representative_predictions
)

prediction_df[
    "prediction_confidence"
] = confidence_values

error_summary_df = (
    prediction_df[
        ~prediction_df[
            "correct_prediction"
        ]
    ]
    .groupby(
        [
            "entity_type",
            "predicted_entity_type"])
    .size()
    .reset_index(
        name="error_count")
    .sort_values(
        "error_count",
        ascending=False))

display(error_summary_df.head(20))
```

## Final interpretation

GraphSAGE achieved a mean macro F1 of 0.111 ± 0.009 on the full graph and 0.321 ± 0.033 on connected-eligible nodes only — a large gain once isolated nodes are excluded, since GraphSAGE can only use neighbourhood information where edges exist.

The canonical graph outperformed the original graph by 0.146 macro F1 under identical Node2Vec and GraphSAGE settings, supporting the value of entity canonicalisation for downstream graph learning.

Training macro F1 rose steadily to ~0.77 while validation macro F1 plateaued around 0.11–0.12, indicating substantial overfitting to the training nodes — expected given the model has far more labelled signal to fit on the training set than the sparse, mostly-isolated test set provides at inference. Early stopping (patience 25) prevented validation performance from degrading further, but did not close the gap. Prediction confidence (mean 0.393) stayed well above the uniform 20-class baseline of 0.05, indicating the model is not just guessing randomly even when accuracy is low.

Limitations: sparse connectivity (only 903 of 4,050 nodes are connected), rare classes excluded from the connected-node experiment, and the canonicalisation counterfactual is pipeline-level rather than a perfectly matched-node causal comparison.
