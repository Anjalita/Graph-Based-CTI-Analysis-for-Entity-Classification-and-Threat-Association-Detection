# 03 Graph Analytics for Cyber Threat Intelligence

## Aim of this Notebook
This notebook performs structural analysis on the Cyber Threat Intelligence Knowledge
Graph constructed in Notebook 2. The purpose is twofold: (1) to understand the graph's
structural properties before applying embeddings/GNNs, and (2) to empirically quantify
the fragmentation problem introduced by mention-level nodes, which directly motivates
the entity canonicalisation performed in Notebook 4.

This stage analyses:
- Graph size and density
- Connected components
- Degree distribution and centrality
- PageRank
- Betweenness centrality
- Important CTI entities
- Community structure

**Note on scope:** because this analysis runs on the raw mention-level graph (before
canonicalisation), community detection and connected-component results below are
expected to be highly fragmented. This is not a bug — it is the evidence that
motivates Notebook 4, and results here are directly compared against the canonical
graph later.

```python

# SECTION 1: PROJECT SETUP

import pickle
from pathlib import Path
from collections import Counter
from networkx.algorithms.community import greedy_modularity_communities

import pandas as pd
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

OUTPUT_DIR= Path("../outputs")

GRAPH_PATH= OUTPUT_DIR/"cti_knowledge_graph.pkl"

print("Graph path:", GRAPH_PATH)
```

### Load Knowledge Graph

```python

# SECTION 2: LOAD KNOWLEDGE GRAPH

with open(GRAPH_PATH,"rb") as file:
    G= pickle.load(file)  #this contains everything after the graph construction

print("Knowledge Graph loaded successfully")
print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
print("Directed:", G.is_directed())
```

### Graph Statistics

```python

# SECTION 3: BASIC GRAPH STATISTICS

num_nodes= G.number_of_nodes()
num_edges= G.number_of_edges()
density= nx.density(G)
self_loops= nx.number_of_selfloops(G)

graph_basic_stats = pd.DataFrame([
    ["Number of nodes", num_nodes],
    ["Number of edges", num_edges],
    ["Graph density", density],
    ["Self-loops", self_loops],
    ["Is directed", G.is_directed()]
], columns=["Metric", "Value"])

display(graph_basic_stats)
```

### Connected Component Analysis

```python

# SECTION 4: CONNECTED COMPONENT ANALYSIS
# to check is there any separate groups becuase all threats need not be connected

weak_components= list(nx.weakly_connected_components(G))

component_sizes= [len(component) for component in weak_components]

component_df = pd.DataFrame({
    "component_id": range(len(component_sizes)),
    "component_size": component_sizes
}).sort_values(by="component_size", ascending=False)

print("Number of weakly connected components:", len(weak_components))
print("Largest component size:", max(component_sizes))
print("Smallest component size:", min(component_sizes))
print(f"Fraction of nodes in components of size 1: "
      f"{sum(1 for s in component_sizes if s == 1) / len(component_sizes):.2%}")

display(component_df.head(20))
```

### Degree Analysis

```python

# SECTION 5: DEGREE ANALYSIS
# used to check how many egdes each node has
degree_dict = dict(G.degree())
in_degree_dict = dict(G.in_degree())
out_degree_dict = dict(G.out_degree())

degree_df = pd.DataFrame({
    "node_id": list(G.nodes()),
    "entity_type": [
        G.nodes[node].get("entity_type")
        for node in G.nodes()],
    "degree": [
        degree_dict[node]
        for node in G.nodes()],
    "in_degree": [
        in_degree_dict[node]
        for node in G.nodes()],
    "out_degree": [
        out_degree_dict[node]
        for node in G.nodes()]})

degree_df = degree_df.sort_values(
    by="degree",
    ascending=False)

display(degree_df.head(20))
```

### Degree Centrality

```python
# SECTION 6: DEGREE CENTRALITY

degree_centrality = nx.degree_centrality(G)

degree_centrality_df = pd.DataFrame({
    "node_id": list(degree_centrality.keys()),
    "degree_centrality": list(degree_centrality.values())})

degree_centrality_df["entity_type"] = degree_centrality_df["node_id"].apply(
    lambda node: G.nodes[node].get("entity_type"))

degree_centrality_df = degree_centrality_df.sort_values(
    by="degree_centrality",
    ascending=False)
display(degree_centrality_df.head(20))
```

```python
# SECTION : DEGREE DISTRIBUTION VISUALISATION

plt.figure(figsize=(8, 5))
plt.hist(degree_df["degree"], bins=50)
plt.yscale("log")
plt.xlabel("Node degree")
plt.ylabel("Number of nodes (log scale)")
plt.title("Degree Distribution of CTI Knowledge Graph")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "degree_distribution.png", dpi=300)
plt.show()
```

### PageRank Analysis

```python

# SECTION 7: PAGERANK ANALYSIS
    # use to differentiate the important connection 
    #for eg if node A has got 10 connections &
        #and node B has 5 conn but connected with those important nodes then 
        #pagerank considerd node B as important


pagerank_scores= nx.pagerank(G,alpha=0.85, max_iter=1000)

pagerank_df = pd.DataFrame({
    "node_id": list(pagerank_scores.keys()),
    "pagerank": list(pagerank_scores.values())})

pagerank_df["entity_type"] = pagerank_df["node_id"].apply(lambda node: G.nodes[node].get("entity_type"))

pagerank_df = pagerank_df.sort_values(by="pagerank",ascending=False)

display(pagerank_df.head(20))
```

### Betweenness Centrality

```python

# SECTION 8: APPROXIMATE BETWEENNESS CENTRALITY
#checked for bridged nodes 

betweenness_scores= nx.betweenness_centrality(G,k=min(500,G.number_of_nodes()),seed=42)

betweenness_df= pd.DataFrame({
    "node_id": list(betweenness_scores.keys()),
    "betweenness_centrality": list(betweenness_scores.values())})

betweenness_df["entity_type"]= betweenness_df["node_id"].apply(
    lambda node: G.nodes[node].get("entity_type"))

betweenness_df= betweenness_df.sort_values(by="betweenness_centrality", ascending=False)
display(betweenness_df.head(20))
```

### Combined Importance Score
Degree centrality and PageRank are weighted equally (0.4 each) as the primary signals
of node importance, since both capture direct and propagated connectivity. Betweenness
centrality is weighted lower (0.2) because on this sparse, fragmented graph most nodes
have zero betweenness by construction, making it a weaker but still useful tie-breaker
for identifying bridge entities. These weights are a design choice, not derived from
optimisation, and are noted as a limitation to revisit if time allows.

```python

# SECTION 9: COMBINED GRAPH IMPORTANCE SCORE
#combining degree centrality+betweeness+pagerank = graph important score 


centrality_df = degree_df.merge(
    degree_centrality_df[["node_id", "degree_centrality"]],
    on="node_id",
    how="left")

centrality_df = centrality_df.merge(
    pagerank_df[["node_id", "pagerank"]],
    on="node_id",
    how="left")

centrality_df = centrality_df.merge(
    betweenness_df[["node_id", "betweenness_centrality"]],
    on="node_id",
    how="left")

for col in ["degree_centrality", "pagerank", "betweenness_centrality"]:
    min_value= centrality_df[col].min()
    max_value=centrality_df[col].max()

    if max_value!= min_value:
        centrality_df[col + "_norm"]= (
            centrality_df[col] - min_value
        ) / (max_value - min_value)
    else:
        centrality_df[col+"_norm"]= 0

centrality_df["graph_importance_score"] = (
    centrality_df["degree_centrality_norm"] * 0.4 +
    centrality_df["pagerank_norm"] * 0.4 +
    centrality_df["betweenness_centrality_norm"] * 0.2
)

centrality_df = centrality_df.sort_values(
    by="graph_importance_score",
    ascending=False
)

display(centrality_df.head(20))
```

### Important Entities by Type

```python

# SECTION 10: IMPORTANT ENTITIES BY TYPE

important_by_type= []

for entity_type in centrality_df["entity_type"].dropna().unique():
    top_entities= centrality_df[
        centrality_df["entity_type"]== entity_type].head(5)
    important_by_type.append(top_entities)

important_by_type_df = pd.concat(
    important_by_type).reset_index(drop=True)

display(important_by_type_df)
```

### Relation Type Distribution in Graph

```python

# SECTION 11: RELATION TYPE DISTRIBUTION IN GRAPH

edge_relation_types= [data.get("relation_type")
    for _, _, data in G.edges(data=True)]
relation_type_graph_counts = Counter(edge_relation_types)

relation_type_graph_df= pd.DataFrame(
    relation_type_graph_counts.items(),
    columns=["relation_type", "count"]).sort_values(by="count", ascending=False)

display(relation_type_graph_df)
```

### COMMUNITY DETECTION

```python
# Community detection works better on undirected graph
#used to identify groups of related cti entities, discover hidden threats, vulns within CTI KG
G_undirected = G.to_undirected()

communities = list(greedy_modularity_communities(G_undirected))

print("Number of communities detected:", len(communities))
print(f"(For reference: the graph has {G.number_of_nodes()} nodes — "
      f"this confirms most 'communities' are single fragmented nodes, "
      f"not meaningful clusters. This motivates canonicalisation in Notebook 4.)")
community_sizes = [len(c) for c in communities]

community_summary_df = pd.DataFrame({
    "community_id": range(len(communities)),
    "number_of_nodes": community_sizes
}).sort_values(by="number_of_nodes", ascending=False)

community_summary_df.head(10)
```

### ASSIGN COMMUNITY ID TO EACH NODE

```python
# SECTION 12: ASSIGN COMMUNITY ID TO EACH NODE

node_community_map = {}

for community_id, community_nodes in enumerate(communities):
    for node in community_nodes:
        node_community_map[node] = community_id

community_df = pd.DataFrame([
    {
        "node_id": node,
        "community_id": community_id
    }
    for node, community_id in node_community_map.items()
])

community_df.head()
```

### COMMUNITY SIZE SUMMARY

```python
# SECTION 13: COMMUNITY SIZE SUMMARY

community_size_df= (
    community_df
    .groupby("community_id")
    .size()
    .reset_index(name="community_size")
    .sort_values(by="community_size", ascending=False))

community_size_df.head(10)
```

### COMMUNITY ENTITY TYPE ANALYSIS

```python
# SECTION 14: COMMUNITY ENTITY TYPE ANALYSIS

# Add node attributes into dataframe
node_type_data = []

for node, data in G.nodes(data=True):
    node_type_data.append({
        "node_id": node,
        "entity_type": data.get("entity_type", "Unknown"),
        "entity_label": data.get("label", data.get("name", "Unknown"))
    })

node_type_df = pd.DataFrame(node_type_data)

community_entity_df= community_df.merge(
    node_type_df,
    on="node_id",
    how="left")

community_entity_summary_df=(
    community_entity_df
    .groupby(["community_id", "entity_type"])
    .size()
    .reset_index(name="count")
    .sort_values(["community_id", "count"], ascending=[True, False]))

community_entity_summary_df.head(20)
```

### TOP ENTITIES INSIDE LARGEST COMMUNITY

```python
# SECTION 15: TOP ENTITIES INSIDE LARGEST COMMUNITY

largest_community_id = community_size_df.iloc[0]["community_id"]

largest_community_nodes = community_df[
    community_df["community_id"] == largest_community_id
]["node_id"].tolist()

largest_community_subgraph = G.subgraph(largest_community_nodes)

print("Largest community ID:", largest_community_id)
print("Nodes in largest community:", largest_community_subgraph.number_of_nodes())
print("Edges in largest community:", largest_community_subgraph.number_of_edges())

largest_degree_df = pd.DataFrame(
    largest_community_subgraph.degree(),
    columns=["node_id", "degree"]).sort_values(by="degree", ascending=False)

largest_degree_df.head(10)
```

### TOP RELATION TYPES IN LARGEST COMMUNITY

```python
# SECTION 16: TOP RELATION TYPES IN LARGEST COMMUNITY

largest_edge_data = []

for source, target, data in largest_community_subgraph.edges(data=True):
    largest_edge_data.append({
        "source": source,
        "target": target,
        "relation_type": data.get("relation_type", "Unknown")})

largest_edge_df = pd.DataFrame(largest_edge_data)

largest_relation_summary_df = (
    largest_edge_df
    .groupby("relation_type")
    .size()
    .reset_index(name="count")
    .sort_values(by="count", ascending=False))

largest_relation_summary_df.head(10)
```

### MERGE COMMUNITY WITH CENTRALITY SCORE

```python
# SECTION 17: MERGE COMMUNITY WITH CENTRALITY SCORE

centrality_community_df = centrality_df.merge(
    community_df,
    on="node_id",
    how="left")

centrality_community_df.head()
```

### TOP IMPORTANT NODES PER COMMUNITY

```python
# SECTION 18: TOP IMPORTANT NODES PER COMMUNITY

top_nodes_per_community_df = (
    centrality_community_df
    .sort_values(["community_id", "graph_importance_score"],ascending=[True, False])
    .groupby("community_id")
    .head(5))

display(top_nodes_per_community_df.head(20))
```

### Output

```python
# ============================================================
# SECTION 19: SAVE GRAPH ANALYTICS OUTPUTS
# ============================================================

graph_basic_stats.to_csv(
    OUTPUT_DIR / "graph_basic_statistics.csv",
    index=False
)

component_df.to_csv(
    OUTPUT_DIR / "graph_component_analysis.csv",
    index=False
)

degree_df.to_csv(
    OUTPUT_DIR / "graph_degree_analysis.csv",
    index=False
)

pagerank_df.to_csv(
    OUTPUT_DIR / "graph_pagerank_scores.csv",
    index=False
)

betweenness_df.to_csv(
    OUTPUT_DIR / "graph_betweenness_scores.csv",
    index=False
)

centrality_df.to_csv(
    OUTPUT_DIR / "graph_combined_importance_scores.csv",
    index=False
)

important_by_type_df.to_csv(
    OUTPUT_DIR / "graph_important_entities_by_type.csv",
    index=False)

relation_type_graph_df.to_csv(
    OUTPUT_DIR / "graph_relation_type_distribution.csv",
    index=False)
# SECTION 18: SAVE COMMUNITY OUTPUTS

community_summary_df.to_csv(OUTPUT_DIR / "community_summary.csv", index=False)
community_df.to_csv(OUTPUT_DIR / "node_communities.csv", index=False)
community_entity_summary_df.to_csv(OUTPUT_DIR / "community_entity_type_summary.csv", index=False)
largest_relation_summary_df.to_csv(OUTPUT_DIR / "largest_community_relation_summary.csv", index=False)
top_nodes_per_community_df.to_csv(OUTPUT_DIR / "top_nodes_per_community.csv", index=False)
print("Community detection outputs saved successfully.")

print("Graph analytics outputs saved successfully.")
```

```python
# SECTION 20: COMMUNITY SIZE BAR CHART

top_communities = community_size_df.head(10)

plt.figure(figsize=(10, 6))
plt.bar(top_communities["community_id"].astype(str), top_communities["community_size"])
plt.xlabel("Community ID")
plt.ylabel("Number of Nodes")
plt.title("Top 10 Largest Communities in CTI Knowledge Graph")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Conclusion

The graph is extremely sparse (density ≈ 0.000007) and highly fragmented: 11,672 weakly connected components across 12,679 nodes, with the largest component holding only 11 nodes. Community detection found a near-identical number of "communities" (~11,672), confirming most are single isolated nodes rather than real clusters — except community 0, which has 11 nodes, is mostly identity entities, and is linked through targets relations.

Of the 1,095 valid relations, only 1,094 appear as graph edges: one pair of entities (report 939) had two relation types (exfiltrates-to and communicates-with) between the same source and target, and the second overwrote the first in the DiGraph.

Degree, PageRank, and betweenness centrality were computed and give a baseline importance ranking, but most nodes score near zero given the fragmentation — only a small set of better-connected entities stand out.

These results motivate the canonicalisation in Notebook 4: resolving repeated mentions into single nodes should reduce this fragmentation and make later graph analytics more meaningful.
