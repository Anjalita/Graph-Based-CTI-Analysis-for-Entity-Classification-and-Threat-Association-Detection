# Node2Vec Graph Embeddings

This notebook converts the final canonical Cyber Threat Intelligence Knowledge Graph
(V4, produced in Notebook 4 via fuzzy-matching canonicalisation) into dense node
embeddings using Node2Vec. It is intentionally focused on representation learning
only; dataset exploration, graph construction, graph analytics, and canonicalisation
are handled in earlier notebooks.

## 1. Aim of this Notebook
The aim of this notebook is to learn structural node representations from the Final
Canonical CTI Knowledge Graph (V4, 4,050 nodes / 910 edges). The resulting embeddings
will be used in Notebook 6 as input features for traditional machine learning
baselines, and later compared with GraphSAGE in Notebook 7.

### Import libraries

This section imports only the libraries required for graph representation learning and fixes random seeds for reproducibility.


```python
import pickle
import time
import random
from pathlib import Path

import numpy as np
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

import torch
import pyg_lib
from torch_geometric.nn import Node2Vec

from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics.pairwise import cosine_similarity

SEED= 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

print("Libraries loaded successfully.")
print("Torch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
print("pyg-lib ready.")
```

### Load Canonical Graph 

Only the final canonical graph is loaded here. Full graph construction and canonicalisation are not repeated.


```python
PROJECT_ROOT= Path("..")
OUTPUT_DIR= PROJECT_ROOT / "outputs"
GRAPH_PATH= OUTPUT_DIR / "cti_knowledge_graph_canonical_final.pkl"

EMBEDDING_DIR= OUTPUT_DIR / "node2vec_embeddings"
FIGURE_DIR= EMBEDDING_DIR / "figures"

EMBEDDING_DIR.mkdir(parents=True, exist_ok=True)
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

with open(GRAPH_PATH, "rb") as f:
    G = pickle.load(f)

print("Canonical graph loaded.")
print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
print("Directed:", G.is_directed())
```

### Node Metadata for Downstream Learning

The node metadata table preserves the mapping between each graph node, its canonical text, entity type, and mention count. This is required for Notebook 6, where `entity_type` becomes the node classification target.


```python
node_list=list(G.nodes())

node_metadata =[]

for node in node_list:
    attrs = G.nodes[node]
    node_metadata.append({
        "node_id": node,
        "entity_type": attrs.get("entity_type"),
        "canonical_text": attrs.get("canonical_text"),
        "mention_count": attrs.get("mention_count", 1),
        "degree": G.degree(node)})

node_metadata_df = pd.DataFrame(node_metadata)

display(node_metadata_df.head())
print("Metadata shape:", node_metadata_df.shape)
print("Unique entity types:", node_metadata_df["entity_type"].nunique())
display(node_metadata_df["entity_type"].value_counts().to_frame("count"))
```

### Convert NetworkX Graph to PyTorch Geometric Format

```python
node_to_idx= {node: idx for idx, node in enumerate(node_list)}
idx_to_node= {idx: node for node, idx in node_to_idx.items()}

node_metadata_df["node_index"] = node_metadata_df["node_id"].map(node_to_idx)

edge_index = torch.tensor([[node_to_idx[src], node_to_idx[dst]] for src, dst in G.edges()], dtype=torch.long).t().contiguous()

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
edge_index = edge_index.to(device)

print("Node mapping complete.")
print("Mapped nodes:", len(node_to_idx))
print("edge_index shape:", tuple(edge_index.shape))
print("Device:", device)
```

### Node2Vec configuration

Node2Vec is used because it learns node representations directly from graph structure using random walks. The first configuration uses unbiased walks (`p = 1`, `q = 1`) to create a neutral baseline before later supervised learning stages. 


| Hyperparameter | Value | Rationale |
|---|---:|---|
| Embedding dimension | 128 | Balanced representation size for a graph with 4,050 nodes |
| Walk length | 20 | Captures local and moderately distant graph neighbourhoods |
| Context size | 10 | Defines the Skip-Gram context window |
| Walks per node | 10 | Improves coverage in a sparse graph |
| p | 1.0 | Neutral return probability |
| q | 1.0 | Neutral inward/outward exploration |
| Epochs | 50 | Sufficient for observing convergence on CPU |


```python

EMBEDDING_DIM= 128
WALK_LENGTH= 20
CONTEXT_SIZE= 10
WALKS_PER_NODE= 10
P= 1.0
Q= 1.0
BATCH_SIZE= 128
LEARNING_RATE = 0.01
EPOCHS = 50

config = {
    "embedding_dim": EMBEDDING_DIM,
    "walk_length": WALK_LENGTH,
    "context_size": CONTEXT_SIZE,
    "walks_per_node": WALKS_PER_NODE,
    "p": P,
    "q": Q,
    "batch_size": BATCH_SIZE,
    "learning_rate": LEARNING_RATE,
    "epochs": EPOCHS,
    "seed": SEED}

display(pd.DataFrame([config]))
```

### Build Node2Vec Model

```python
model = Node2Vec(
    edge_index=edge_index,
    embedding_dim=EMBEDDING_DIM,
    walk_length=WALK_LENGTH,
    context_size=CONTEXT_SIZE,
    walks_per_node=WALKS_PER_NODE,
    p=P,
    q=Q,
    sparse=True
).to(device)

loader = model.loader(batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
optimizer = torch.optim.SparseAdam(model.parameters(), lr=LEARNING_RATE)

print("Node2Vec model created successfully.")
print(model)
```

### Train Node2Vec


Training is monitored using the Node2Vec loss. The loss curve is saved so that convergence can be discussed in the dissertation.


```python
def train_node2vec():
    model.train()
    total_loss = 0

    for pos_rw, neg_rw in loader:
        optimizer.zero_grad()
        pos_rw = pos_rw.to(device)
        neg_rw = neg_rw.to(device)
        loss = model.loss(pos_rw, neg_rw)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    return total_loss / len(loader)
```

```python
# SECTION: RUN TRAINING (start with 1 epoch to sanity-check timing first)
loss_history = []
start_time= time.time()

for epoch in range(1, EPOCHS + 1):
    loss= train_node2vec()
    loss_history.append(loss)

    if epoch == 1 or epoch % 5 == 0:
        print(f"Epoch {epoch:03d} | Loss: {loss:.4f}")

training_time = time.time() - start_time

print("Training complete.")
print("Final loss:", round(loss_history[-1], 4))
print("Training time:", round(training_time, 2), "seconds")
```

### Plot Training Loss

```python
plt.figure(figsize=(8, 5))
plt.plot(range(1, EPOCHS + 1), loss_history)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Node2Vec Training Loss")
plt.grid(True)
plt.tight_layout()
plt.savefig(FIGURE_DIR / "node2vec_training_loss.png", dpi=300)
plt.show()
```

### Generate Node Embeddings

```python
model.eval()

with torch.no_grad():
    embeddings = model.embedding.weight.detach().cpu().numpy()

print("Embeddings generated.")
print("Shape:", embeddings.shape)
```

### Create Embedding DataFrame

```python
embedding_cols= [f"emb_{i}" for i in range(EMBEDDING_DIM)]

embeddings_df = pd.DataFrame(embeddings, columns=embedding_cols)
embeddings_df.insert(0, "node_index", range(len(embeddings_df)))
embeddings_df.insert(1, "node_id", [idx_to_node[i] for i in range(len(embeddings_df))])

embeddings_df = embeddings_df.merge(node_metadata_df,on=["node_index", "node_id"],how="left")

display(embeddings_df.head())
print("Embedding dataframe shape:", embeddings_df.shape)
```

### Embedding Quality Check: Vector Statistics

This lightweight check verifies that the embedding vectors were generated correctly and are numerically usable for downstream machine learning.


```python
embedding_norms= np.linalg.norm(embeddings, axis=1)

embedding_quality_summary = pd.DataFrame([{
    "num_nodes": embeddings.shape[0],
    "embedding_dim": embeddings.shape[1],
    "mean_value": float(np.mean(embeddings)),
    "std_value": float(np.std(embeddings)),
    "min_value": float(np.min(embeddings)),
    "max_value": float(np.max(embeddings)),
    "mean_l2_norm": float(np.mean(embedding_norms)),
    "std_l2_norm": float(np.std(embedding_norms))
}])

display(embedding_quality_summary)
```

### Nearest-Neighbour Embedding Check

Nearest-neighbour analysis provides an interpretable check of whether structurally related cyber entities occupy nearby positions in the learned embedding space. A high-degree node is used as the example query because isolated nodes do not provide meaningful random-walk context.


```python
def find_nearest_nodes(query_node, top_k=10):
    if query_node not in node_to_idx:
        raise ValueError(f"Node not found: {query_node}")

    query_idx= node_to_idx[query_node]
    query_vec= embeddings[query_idx].reshape(1, -1)

    similarities= cosine_similarity(query_vec, embeddings)[0]
    nearest_indices= similarities.argsort()[::-1][1:top_k + 1]

    results= []

    for idx in nearest_indices:
        node_id = idx_to_node[idx]
        attrs = G.nodes[node_id]

        results.append({
            "query_node": query_node,
            "nearest_node": node_id,
            "similarity": similarities[idx],
            "entity_type": attrs.get("entity_type"),
            "canonical_text": attrs.get("canonical_text"),
            "degree": G.degree(node_id)})

    return pd.DataFrame(results)


degree_dict= dict(G.degree())
example_node= max(degree_dict, key=degree_dict.get)

nearest_df= find_nearest_nodes(example_node, top_k=10)

print("Query node:", example_node)
print("Query degree:", G.degree(example_node))
display(nearest_df)
```

### Quantitative embedding evaluation

```python
# SECTION: QUANTITATIVE EMBEDDING EVALUATION — LINK PREDICTION
from sklearn.metrics import roc_auc_score, average_precision_score

# Held-out positive edges (sample of real edges)
rng= np.random.RandomState(SEED)
real_edges= list(G.edges())
n_test= max(1, int(0.2 * len(real_edges)))
test_pos_edges = [real_edges[i] for i in rng.choice(len(real_edges), n_test, replace=False)]

# Negative edges: random non-existent pairs
existing_edges = set(G.edges())
test_neg_edges = []
nodes_list = list(G.nodes())
while len(test_neg_edges) < n_test:
    u, v = rng.choice(nodes_list, 2, replace=False)
    if (u, v) not in existing_edges and (v, u) not in existing_edges:
        test_neg_edges.append((u, v))

def edge_score(u, v):
    return float(np.dot(embeddings[node_to_idx[u]], embeddings[node_to_idx[v]]))

pos_scores= [edge_score(u, v) for u, v in test_pos_edges]
neg_scores= [edge_score(u, v) for u, v in test_neg_edges]

y_true= [1]* len(pos_scores) + [0] * len(neg_scores)
y_scores= pos_scores + neg_scores

link_pred_auc = roc_auc_score(y_true, y_scores)
link_pred_ap = average_precision_score(y_true, y_scores)

print(f"Link prediction ROC-AUC: {link_pred_auc:.4f}")
print(f"Link prediction Average Precision: {link_pred_ap:.4f}")

pd.DataFrame([{
    "metric": "ROC-AUC", "value": link_pred_auc
}, {
    "metric": "Average Precision", "value": link_pred_ap
}]).to_csv(EMBEDDING_DIR / "node2vec_link_prediction_evaluation.csv", index=False)
```

### PCA Visualisation
PCA provides a linear two-dimensional projection of the embedding space. It is used as a quick global overview of how entity types are distributed after representation learning.

```python
pca = PCA(n_components=2, random_state=SEED)
pca_result = pca.fit_transform(embeddings)

pca_df = pd.DataFrame({
    "x": pca_result[:, 0],
    "y": pca_result[:, 1],
    "entity_type": node_metadata_df["entity_type"].values,
    "node_id": node_metadata_df["node_id"].values})

plt.figure(figsize=(10, 7))

for entity_type in pca_df["entity_type"].dropna().unique():
    subset = pca_df[pca_df["entity_type"] == entity_type]
    plt.scatter(subset["x"], subset["y"], s=10, alpha=0.6, label=entity_type)

plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.title("PCA Visualisation of Node2Vec Embeddings")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left", fontsize=8)
plt.tight_layout()
plt.savefig(FIGURE_DIR / "node2vec_pca_visualisation.png", dpi=300)
plt.show()

print("Explained variance ratio:", pca.explained_variance_ratio_)
```

### t-SNE Visualisation
t-SNE is used as a complementary non-linear visualisation to inspect local neighbourhood patterns. A sample is used to keep the computation efficient.


```python
sample_size = min(1500, len(embeddings))
sample_indices = np.random.RandomState(SEED).choice(
    len(embeddings),
    sample_size,
    replace=False)

tsne = TSNE(
    n_components=2,
    perplexity=30,
    learning_rate="auto",
    init="pca",
    random_state=SEED)

tsne_result = tsne.fit_transform(embeddings[sample_indices])

tsne_df = pd.DataFrame({
    "x": tsne_result[:, 0],
    "y": tsne_result[:, 1],
    "entity_type": node_metadata_df.iloc[sample_indices]["entity_type"].values,
    "node_id": node_metadata_df.iloc[sample_indices]["node_id"].values})

plt.figure(figsize=(10, 7))

for entity_type in tsne_df["entity_type"].dropna().unique():
    subset = tsne_df[tsne_df["entity_type"] == entity_type]
    plt.scatter(subset["x"], subset["y"], s=10, alpha=0.6, label=entity_type)

plt.xlabel("t-SNE Dimension 1")
plt.ylabel("t-SNE Dimension 2")
plt.title("t-SNE Visualisation of Node2Vec Embeddings")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left", fontsize=8)
plt.tight_layout()
plt.savefig(FIGURE_DIR / "node2vec_tsne_visualisation.png", dpi=300)
plt.show()
```

### Output

```python
# ============================================================
# Save Notebook 5 outputs
# ============================================================

np.save(EMBEDDING_DIR / "node2vec_embeddings.npy",embeddings)

embeddings_df.to_csv(EMBEDDING_DIR / "node2vec_embeddings_with_metadata.csv",index=False)

node_metadata_df.to_csv(EMBEDDING_DIR / "node_metadata.csv",index=False)

pd.DataFrame({
    "node_id": list(node_to_idx.keys()),
    "node_index": list(node_to_idx.values())}).to_csv(EMBEDDING_DIR / "node_index_mapping.csv",index=False)

pd.DataFrame({
    "epoch": range(1, EPOCHS + 1),
    "loss": loss_history}).to_csv(EMBEDDING_DIR / "node2vec_loss_history.csv",index=False)

training_summary = pd.DataFrame([{
    "graph_version": "canonical_v4_final",
    "num_nodes": G.number_of_nodes(),
    "num_edges": G.number_of_edges(),
    "embedding_dim": EMBEDDING_DIM,
    "walk_length": WALK_LENGTH,
    "context_size": CONTEXT_SIZE,
    "walks_per_node": WALKS_PER_NODE,
    "p": P,
    "q": Q,
    "batch_size": BATCH_SIZE,
    "learning_rate": LEARNING_RATE,
    "epochs": EPOCHS,
    "seed": SEED,
    "final_loss": loss_history[-1],
    "training_time_seconds": training_time,
    "link_prediction_auc": link_pred_auc,
    "link_prediction_ap": link_pred_ap}])

training_summary.to_csv(EMBEDDING_DIR / "node2vec_training_summary.csv",index=False)

embedding_quality_summary.to_csv(EMBEDDING_DIR / "node2vec_embedding_quality_summary.csv",index=False)

nearest_df.to_csv( EMBEDDING_DIR / "node2vec_nearest_neighbour_example.csv",index=False)

link_prediction_evaluation = pd.DataFrame([{"roc_auc": link_pred_auc,"average_precision": link_pred_ap}])

link_prediction_evaluation.to_csv(EMBEDDING_DIR / "node2vec_link_prediction_evaluation.csv",index=False)

torch.save(model.state_dict(),EMBEDDING_DIR / "node2vec_model_state_dict.pt")

print("Notebook 5 outputs saved successfully.")
print("Output folder:", EMBEDDING_DIR.resolve())
print("Saved trained Node2Vec model state.")
print(f"ROC-AUC: {link_pred_auc:.4f}")
print(f"Average Precision: {link_pred_ap:.4f}")
```
