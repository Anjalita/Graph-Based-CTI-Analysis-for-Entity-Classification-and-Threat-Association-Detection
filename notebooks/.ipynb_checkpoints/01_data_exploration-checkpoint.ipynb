# 01 Data Exploration — Cyber Threat Intelligence Knowledge Graph

## Aim of this Notebook
This notebook explores the CTI dataset to assess its suitability for knowledge graph
construction. It validates dataset structure, quantifies entity/relation coverage,
and identifies the entity and relation vocabulary used to design the graph schema
in Notebook 2.

## Project Plan
1. Project setup and file inventory
2. Load and cross-check JSONL/CSV dataset variants
3. Structural and missing-value assessment
4. Entity analysis (types, frequency, coverage)
5. Relation analysis (types, frequency, coverage)
6. Data quality checks relevant to downstream graph construction


```python

# section 1: PROJECT SETUP


import os
import ast
import json
import pickle
import warnings
from pathlib import Path
from collections import Counter, defaultdict

import pandas as pd
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

PROJECT_ROOT = Path("..")    #to find my dataset folder
DATA_DIR = PROJECT_ROOT / "data" / "archive"
OUTPUT_DIR = PROJECT_ROOT / "outputs"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


#to store all the files
DATA_FILES = {
    "all_jsonl": DATA_DIR / "all.jsonl",
    "train_jsonl": DATA_DIR / "train.jsonl",
    "validation_jsonl": DATA_DIR / "validation.jsonl",
    "test_jsonl": DATA_DIR / "test.jsonl",

    "all_csv": DATA_DIR / "cyber-threat-intelligence_all.csv",
    "train_csv": DATA_DIR / "cyber-threat-intelligence-splited_train.csv",
    "validation_csv": DATA_DIR / "cyber-threat-intelligence-splited_validate.csv",
    "test_csv": DATA_DIR / "cyber-threat-intelligence-splited_test.csv",

    "processed_csv": DATA_DIR / "Cyber-Threat-Intelligence-Custom-Data_new_processed.csv"
}

print("Project root:", PROJECT_ROOT.resolve())
print("Data directory:", DATA_DIR.resolve())
print("Output directory:", OUTPUT_DIR.resolve())
```

```python

# Section 2: DATASET INVENTORY


def file_inventory(file_dict):
    records = []

    for name, path in file_dict.items():
        records.append({
            "dataset_name": name,
            "path": str(path),
            "exists": path.exists(),
            "file_size_kb": round(path.stat().st_size / 1024, 2) if path.exists() else None
        })

    return pd.DataFrame(records)

inventory_df = file_inventory(DATA_FILES)
display(inventory_df)

```

```python

# Section 3: LOAD Original CTI DATASETS


def load_jsonl(path):
    records = []

    with open(path, "r", encoding="utf-8") as file:
        for line in file:
            if line.strip():
                records.append(json.loads(line))

    return pd.DataFrame(records)


df_all = load_jsonl(DATA_FILES["all_jsonl"])    #loads full cti dataset
#df_train, validate and test will loads split datasets
df_train = load_jsonl(DATA_FILES["train_jsonl"])
df_validation = load_jsonl(DATA_FILES["validation_jsonl"])
df_test = load_jsonl(DATA_FILES["test_jsonl"])

print("All dataset:", df_all.shape)
print("Train dataset:", df_train.shape)
print("Validation dataset:", df_validation.shape)
print("Test dataset:", df_test.shape)

display(df_all.head()) 
```

```python

# SECTION 4: LOAD CSV DATASETS FOR CROSS-CHECKING


df_all_csv = pd.read_csv(DATA_FILES["all_csv"])
df_train_csv = pd.read_csv(DATA_FILES["train_csv"])
df_validation_csv = pd.read_csv(DATA_FILES["validation_csv"])
df_test_csv = pd.read_csv(DATA_FILES["test_csv"])
df_processed = pd.read_csv(DATA_FILES["processed_csv"])

# Dataset comparison
dataset_summary = pd.DataFrame({
    "dataset": [
        "all.jsonl",
        "train.jsonl",
        "validation.jsonl",
        "test.jsonl",
        "all_csv",
        "train_csv",
        "validation_csv",
        "test_csv",
        "processed_csv"
    ],
    "rows": [
        len(df_all),
        len(df_train),
        len(df_validation),
        len(df_test),
        len(df_all_csv),
        len(df_train_csv),
        len(df_validation_csv),
        len(df_test_csv),
        len(df_processed)
    ],
    "columns": [
        df_all.shape[1],
        df_train.shape[1],
        df_validation.shape[1],
        df_test.shape[1],
        df_all_csv.shape[1],
        df_train_csv.shape[1],
        df_validation_csv.shape[1],
        df_test_csv.shape[1],
        df_processed.shape[1]
    ]
})

display(dataset_summary)
display(df_processed.head())
```

```python

# SECTION 5: DATASET ROLE DEFINITION

dataset_roles = pd.DataFrame([
    {
        "dataset": "all.jsonl",
        "role": "Main complete CTI dataset",
        "use": "Knowledge Graph construction and global entity/relation analysis"
    },
    {
        "dataset": "train.jsonl",
        "role": "Training split",
        "use": "Model training and baseline fitting"
    },
    {
        "dataset": "validation.jsonl",
        "role": "Validation split",
        "use": "Model tuning and threshold selection"
    },
    {
        "dataset": "test.jsonl",
        "role": "Test split",
        "use": "Final model evaluation"
    },
    {
        "dataset": "cyber-threat-intelligence_all.csv",
        "role": "CSV version of CTI dataset",
        "use": "Cross-checking dataset structure and entity distribution"
    },
    {
        "dataset": "Cyber-Threat-Intelligence-Custom-Data_new_processed.csv",
        "role": "Processed enriched dataset",
        "use": "Diagnosis and solution analysis"
    }
])

display(dataset_roles)

dataset_roles.to_csv(
    OUTPUT_DIR / "dataset_role_definition.csv",
    index=False
)
```

```python

# SECTION 6.1: BASIC DATASET STRUCTURE
# 

print("Rows:", df_all.shape[0])
print("Columns:", df_all.shape[1])

print("\nColumns:")
for col in df_all.columns:
    print("-", col)
```

```python
# SECTION 6.2: MISSING VALUES
missing_summary = df_all.isnull().sum().reset_index()
missing_summary.columns = ["column", "missing_count"]
missing_summary["missing_pct"] = (
    missing_summary["missing_count"] / len(df_all) * 100
).round(2)
display(missing_summary)
```

```python

# SECTION 6.3: REPORTS WITH ENTITIES and RELATION

coverage_summary = pd.DataFrame([
    {"metric": "Reports with entities",
     "count": df_all["entities"].apply(lambda x: len(x) > 0).sum(),
     "pct_of_total": round(df_all["entities"].apply(lambda x: len(x) > 0).mean() * 100, 2)},
    {"metric": "Reports with relations",
     "count": df_all["relations"].apply(lambda x: len(x) > 0).sum(),
     "pct_of_total": round(df_all["relations"].apply(lambda x: len(x) > 0).mean() * 100, 2)},
])
display(coverage_summary)
```

```python
# SECTION 6.5: TEXT LENGTH ANALYSIS
df_all["text_length"] = df_all["text"].str.len()

print("Shortest report:", df_all["text_length"].min())
print("Longest report:", df_all["text_length"].max())
print("Average report length:", round(df_all["text_length"].mean(), 2))

plt.figure(figsize=(8, 5))
plt.hist(df_all["text_length"], bins=50)
plt.xlabel("Report text length (characters)")
plt.ylabel("Number of reports")
plt.title("Distribution of CTI Report Lengths")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "report_length_distribution.png", dpi=300)
plt.show()
```

# Entity Analysis

```python
# SECTION 7.1: TOTAL ENTITY COUNT
# its about how many times each entity has occured?
# Reusable helper — defined once, used in entity + quality-check sections
def extract_entity_labels(df):
    labels = []
    for entity_list in df["entities"]:
        for entity in entity_list:
            labels.append(entity["label"])
    return labels

entity_labels = extract_entity_labels(df_all)
print("Total extracted entity labels:", len(entity_labels))
```

```python
# SECTION 7.3-7.4: ENTITY TYPE DISTRIBUTION
entity_type_counts = pd.Series(entity_labels).value_counts()
display(entity_type_counts.head(20))

plt.figure(figsize=(10, 6))
entity_type_counts.head(20).plot(kind="bar")
plt.ylabel("Count")
plt.title("Top 20 Entity Types in CTI Dataset")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "entity_type_distribution.png", dpi=300)
plt.show()
```

# Relation Analysis

```python

# SECTION 8.1: TOTAL RELATION COUNT

def extract_relation_types(df):
    types = []
    for relation_list in df["relations"]:
        for relation in relation_list:
            if "type" in relation:
                types.append(relation["type"])
    return types

relation_types = extract_relation_types(df_all)
print("Total extracted relation labels:", len(relation_types))

relation_type_counts = pd.Series(relation_types).value_counts()
display(relation_type_counts)

plt.figure(figsize=(10, 6))
relation_type_counts.plot(kind="bar")
plt.ylabel("Count")
plt.title("Relation Type Distribution in CTI Dataset")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "relation_type_distribution.png", dpi=300)
plt.show()
```

# Data Quality Assessment and Cleaning

```python
# SECTION 9.0: STRUCTURAL VALIDITY CHECKS
duplicate_reports = df_all["id"].duplicated().sum()
print("Duplicate report IDs:", duplicate_reports)

empty_text_reports = (df_all["text"].str.len() == 0).sum()
print("Reports with empty text:", empty_text_reports)

# Validate entity offsets against report text length (critical for Notebook 4)
invalid_offset_count = 0
for _, row in df_all.iterrows():
    text_len = len(row["text"])
    for entity in row["entities"]:
        start, end = entity.get("start_offset"), entity.get("end_offset")
        if start is None or end is None or start < 0 or end > text_len or start >= end:
            invalid_offset_count += 1

print("Entities with invalid offsets:", invalid_offset_count)
```

```python
# SECTION 9.1: UNIQUE ENTITY TYPES
unique_entity_types = sorted(set(entity_labels))
print("Number of unique entity types:", len(unique_entity_types))
for entity_type in unique_entity_types:
    print(entity_type)
```

```python

# SECTION 9.2: UNIQUE RELATION TYPES


unique_relation_types = sorted(
    set(relation_types)
)

print("Number of unique relation types:",
      len(unique_relation_types))

for relation_type in unique_relation_types:
    print(relation_type)
```

## Conclusion
The dataset contains 9,732 reports with 20 unique entity types and 25 unique relation
types. Only a minority of reports contain annotated relations, and entity/relation
density per report is low. This sparsity is carried forward into the knowledge graph
built in Notebook 2 and is a key limitation addressed later through canonicalisation
(Notebook 4).
